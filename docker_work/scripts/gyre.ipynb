```python
# Import modules

import pygyre as pg
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress
```

```python
# Read data from a gyre summary file

s = pg.read_output('../lesson4/gyre_out/gyre_summary3.txt')
s
```

```python
# Plot the data

plt.figure()

plt.plot(s['n_pg'], 1/s['freq'].real)

plt.xlabel('n_pg')
plt.ylabel('Period (day)')
plt.ylim(0, 4)

plt.savefig("../plots/gmode3.png", bbox_inches='tight')
plt.show()
```

```python
res = linregress(s['n_pg'], 1/s['freq'].real)
(res.slope * 24 + 1) * 60
```
